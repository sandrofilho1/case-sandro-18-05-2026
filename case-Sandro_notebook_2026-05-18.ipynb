````python
from pyspark.sql.functions import countDistinct, collect_set, count

df_resumo = df_total.groupBy("User").agg(
    count("*").alias("total_registros"),
    collect_set("Model").alias("modelos_utilizados"),
    collect_set("Device").alias("dispositivos"),
    collect_set("gt").alias("atividades_realizadas"),
    collect_set("fonte").alias("fontes_de_dados")
)

df_resumo = df_resumo.orderBy("User")
display(df_resumo)
````

````python
from pyspark.sql.functions import from_unixtime, col

df_total_convertido = df_total \
    .withColumn(
        "Arrival_Time_legivel",
        from_unixtime(col("Arrival_Time") / 1000)
    ) \
    .withColumn(
        "Creation_Time_legivel",
        from_unixtime(col("Creation_Time") / 1000000000)
    )

df_total_convertido.select(
    "User",
    "Arrival_Time",
    "Arrival_Time_legivel",
    "Creation_Time",
    "Creation_Time_legivel"
).show(5, truncate=False)
````

````python
from pyspark.sql.functions import lag, col
from pyspark.sql.window import Window

# Define a janela: agrupa por User e Device, ordena por Arrival_Time
janela = Window.partitionBy("User", "Device").orderBy("Arrival_Time")

# Calcula o intervalo em milissegundos
df_com_intervalo = df_total_convertido \
    .withColumn(
        "Arrival_Time_anterior",
        lag("Arrival_Time").over(janela)
    ) \
    .withColumn(
        "intervalo_ms",
        col("Arrival_Time") - col("Arrival_Time_anterior")
    )

# Mostra o resultado
df_com_intervalo.select(
    "User",
    "Device",
    "Arrival_Time_legivel",
    "Arrival_Time_anterior",
    "intervalo_ms"
).show(10, truncate=False)
````

````python
from pyspark.sql.functions import avg, min as spark_min, max as spark_max, round as spark_round

df_com_intervalo.groupBy("User", "Device") \
    .agg(
        spark_round(avg("intervalo_ms"), 2).alias("intervalo_medio_ms"),
        spark_min("intervalo_ms").alias("intervalo_minimo_ms"),
        spark_max("intervalo_ms").alias("intervalo_maximo_ms")
    ) \
    .orderBy("User", "Device") \
    .show(20, truncate=False)
````

## Tarefa 7 — Análise de performance e estratégias de otimização

### Problemas identificados no código atual

**1. count() múltiplos vezes**
Cada chamada de `count()` relê os arquivos do zero.
Com 34 milhões de registros e arquivos de 3 GB, isso é custoso.

**2. Operações de Window (lag) sem particionamento eficiente**
A função `lag` ordenou 34 milhões de linhas — uma das operações
mais pesadas no Spark.

**3. Arquivos CSV são lentos**
CSV não tem compressão nem índices — o Spark precisa ler
byte a byte.

### Estratégias de otimização

**1. Converter CSV para Delta Lake ou Parquet**
Parquet é colunar e comprimido — leitura até 10x mais rápida.
```python
df_total.write.format("delta").save("/Volumes/workspace/default/dados_delta/")
```

**2. Usar cache() para DataFrames reutilizados**
Evita reler os arquivos a cada operação.
```python
df_total.cache()
df_total.count()  # força o cache na memória
```

**3. Reparticionar os dados**
Distribuir melhor os dados entre os nós de processamento.
```python
df_total = df_total.repartition(8, "User", "Device")
```

**4. Evitar collect_set em dados muito grandes**
`collect_set` traz dados para o driver — perigoso em escala.
Para datasets maiores, preferir `countDistinct`.

**5. Usar predicate pushdown**
Filtrar antes de processar — o Spark lê menos dados.
```python
# Em vez de filtrar depois, filtrar ao carregar
df_phone_acc = spark.read.csv(...).filter(col("gt") != "null")
```
